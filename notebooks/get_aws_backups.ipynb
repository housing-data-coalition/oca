```python
# notebook used after restoring from a .dump

# When sql dump fails, grab zips from s3 backup
import dotenv, re, os
from importlib.machinery import SourceFileLoader
S3 = SourceFileLoader("S3", "/app/lib/s3.py").load_module().S3

dotenv.load_dotenv('/app/.env')
S3_PRIVATE_FOLDER = 'private'
priv_dir = os.path.abspath('/app/lib/data-private')
```

```python
aws_id = os.environ.get('AWS_ACCESS_KEY_ID', '')
aws_key = os.environ.get('AWS_SECRET_ACCESS_KEY', '')
aws_bucket_name = 'oca-2-dev'
```

```python
backup_s3 = S3(aws_id, aws_key, aws_bucket_name)
```

```python
backups = backup_s3.list_files('.+\.zip', S3_PRIVATE_FOLDER)
```

```python
# filter for everything after oca_03_17_2024.dump
def sort_by_date(file):
    r = re.search(r'(\d+.+)\.zip', file).group(0).replace('.',' ')
    return r

backups = sorted(backups, key = sort_by_date)
```

```python
after_dump_backups = backups[backups.index('LandlordTenant.Incr.2024-03-24.zip'):]
len(after_dump_backups)
```

```python
for f in after_dump_backups:
    backup_s3.download_file(f"{S3_PRIVATE_FOLDER}/{f}", os.path.join(priv_dir, f))
```
